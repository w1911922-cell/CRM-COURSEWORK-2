```python
# numpy and pandas will be used for data manipulation
import numpy as np
import pandas as pd

#matplotib will be used for visually representing our data
import matplotlib.pyplot as plt
#yfinance will be used for importing historical oil prices
import yfinance as yfin

```

```python
#Task A Question A
ticker = 'AMZN'
ticker = yfin.Ticker(ticker)

#importing our data
data = ticker.history(period='5y')
```

```python
data
```

```python
# Setting the text on the Y-axis
plt.ylabel("Amazon")

# Setting the size of our graph
data ['Close'].plot(figsize=(10,5))
```

```python
#Task A Question B
data ['MA_20'] = data ['Close'].rolling(window=20).mean()
data ['MA_50'] = data ['Close'].rolling(window=50).mean()

# Calculate Exponential Moving Average (EMA)
data ['EMA_20'] = data ['Close'].ewm(span=20, adjust=False).mean()

# Calculate Relative Strength Index (RSI)
delta = data ['Close'].diff(1)
gain = delta.where(delta > 0, 0)
loss = -delta.where(delta < 0, 0)
avg_gain = gain.rolling(window=14).mean()
avg_loss = loss.rolling(window=14).mean()
rs = avg_gain / avg_loss
data ['RSI'] = 100 - (100 / (1 + rs))

data
```

```python
# Drop rows with NaN values
data = data.dropna()

data
```

```python
# Initialising X and assigning the two feature variables
X = data [['EMA_20', 'MA_50']]

# Getting the head of the data
X.head()
```

```python
 # Setting-up the dependent variable
y = data ['Close']

 # Getting the head of the data
y.head()
```

```python
#Task A Question C

#Setting the training set to 70% of the data

training = 0.7
t = int(training * len(data))

# Training dataset
X_train = X[:t]
y_train = y[:t]

# Testing dataset
X_test = X[t:]
y_test = y[t:]
```

```python
#Reproducible Seed
import numpy as np
np.random.seed(42)
```

```python
#Task A Question D
from sklearn.linear_model import LinearRegression
model = LinearRegression().fit(X_train,y_train)
print("Linear Regression Model Defined and Trained")
```

```python
#Task A Question E
predicted_price = model.predict(X_test)
predicted_price = pd.DataFrame(predicted_price, index=y_test.index, columns=['Close'])
predicted_price.plot(figsize=(10,5))
y_test.plot()
plt.legend(['Predicted Price', 'Actual Price'])
plt.ylabel("Amazon Stock Price")
plt.show()
```

```python
# Computing the accuracy of our model
R_squared_score = model.score(X[t:], y[t:])*100
accuracy = ("{0:.2f}").format(R_squared_score)
print ("The model has a " + accuracy + "% accuracy.")
```

```python
#Task A Question f

#Extract alpha & beta values
print(f'alpha = {model.intercept_}')
print(f'betas = {model.coef_}')
```

```python
#Formulate Linear Regression Model
alpha = model.intercept_
beta1, beta2 = model.coef_

sign = "-" if beta2 < 0 else "+"

print(f"Linear Regression Model: ŷ = {alpha:.4f} + {beta1:.4f}(EMA_20) {sign} {abs(beta2):.4f}(MA_50)")
```
